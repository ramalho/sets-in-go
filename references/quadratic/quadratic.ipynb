# Quadratic vs. linear growth

How the running time of an $O(n^2)$ algorithm pulls away from an $O(n)$ one as the input grows.

The hand-drawn look needs the *xkcd* font; without it matplotlib falls back to a plain sans-serif. Install it with:

```
curl -L -o ~/.local/share/fonts/xkcd.otf \
  https://github.com/ipython/xkcd-font/raw/master/xkcd/build/xkcd.otf
fc-cache -f && rm -rf ~/.cache/matplotlib
```

````python
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
````

````python
N = 20
n = np.linspace(0, N, 400)
quadratic = n**2
linear = n

RED = '#d62728'
BLUE = '#1f77b4'
````

````python
with plt.xkcd():
    # the xkcd style lists fonts we don't have; keep only the one we installed,
    # and use it for math text too, so the superscript matches the rest
    plt.rcParams['font.family'] = 'xkcd'
    plt.rcParams['mathtext.fontset'] = 'custom'
    for face in ('rm', 'it', 'bf', 'sf', 'cal', 'tt'):
        plt.rcParams[f'mathtext.{face}'] = 'xkcd'

    fig, ax = plt.subplots(figsize=(8, 6))

    ax.plot(n, quadratic, color=RED, linewidth=2) #, label='$O(n^2)$')
    ax.plot(n, linear, color=BLUE, linewidth=2) #, label='$O(n)$')

    ax.annotate('$O(n^2)$', xy=(N, N**2), xytext=(-70, -30),
               textcoords='offset points', color=RED, fontsize=16)
    ax.annotate('$O(n)$', xy=(N, N), xytext=(-60, 20),
                textcoords='offset points', color=BLUE, fontsize=16)

    ax.set_xlim(0, N * 1.05)
    ax.set_ylim(0, N**2 * 1.05)
    ax.set_xlabel('input size (n)')
    ax.set_ylabel('operations')
    ax.set_title('Quadratic vs. linear growth')
    ax.set_xticks([0, N // 2, N])
    ax.set_yticks([0, N**2 // 2, N**2])
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    # ax.legend(loc='upper left', frameon=False, fontsize=14)

    fig.tight_layout()
    fig.savefig('quadratic.png', dpi=150)  # 8x6 inches at 150 dpi = 1200x900
    plt.show()
````
